5. Дана информация о времени заезда и отъезда посетителей отеля. Необходимо определить, в какой день посетителей в отеле единомоментно находилось больше всего.
Пример входных данных (один элемент данного листа – кортеж, содержащий дату заезда и отъезда одного посетителя): [(“2024-09-15”, “2024-09-15”), (“2024-09-14”, “2024-09-21”)]

```python
#мой способ
import pandas as pd

def longest_data(values):
    
    datas = {} #ключ - число, значение - число человеков
    for i in values:
        for date in  pd.date_range(start = i[0], end = i[1]):
            if date.date() in datas:
                datas[date.date()] += 1
            else:
                datas[date.date()] = 1
    print(datas.items())
    return(max(datas, key=datas.get))

print(longest_data([('2024-09-15', '2024-09-15'), ('2024-09-14', '2024-09-21')]))

```

```python
# по посказкам ИИ: метод сканирующей прямой
def longest_data_2(values):
    dates = []
    for i in values:
        dates.append((i[0],1))
        dates.append((i[1],-1))
    dates.sort()
    max_amount = 0
    cur_amount = 0
    max_date = None

    for date in dates:
        max_date = dates[0][0]
        cur_amount += date[1]

        if cur_amount > max_amount:
            max_amount = cur_amount
            max_date = date[0]

    return max_date, max_amount
```
